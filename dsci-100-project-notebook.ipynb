```r
numeric_summary <- basket_ball_useful |>
  pivot_longer(where(is.numeric), names_to = "variable", values_to = "value") |>
  group_by(variable) |>
  summarise(
    mean = mean(value),
    min  = min(value),
    max  = max(value)
  )

numeric_summary

#I found where(is.numeric) on google to only select for numericcal values
```
